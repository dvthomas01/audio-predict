```python
import torch
import torch.nn as nn
from typing import Tuple
from ../models.lstm_baseline import LSTMBaseline, count_parameters
```

```python
# Test the model
batch_size = 4
freq_bins = 128
context_size = 10
prediction_horizon = 5

model = LSTMBaseline(
    input_dim=freq_bins,
    hidden_dim=256,
    num_layers=2
)

# 1. Test Training Mode (Sequence-to-Sequence)
x = torch.randn(batch_size, freq_bins, context_size)
output, _ = model(x)
print(f"Training Input:  {x.shape}")
print(f"Training Output: {output.shape}") # Should match input

# 2. Test Inference Mode (Autoregressive)
future = model.predict_autoregressive(x, prediction_horizon)
print(f"Inference Output: {future.shape}") # Should be (batch, freq, horizon)
print(f"Model parameters: {count_parameters(model):,}")
```
